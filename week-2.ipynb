### Basic statical analysis

```python
sns.countplot(data=df,x="Species")

plt.title("Distribution of Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
numerical_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

df[numerical_columns].hist(figsize=(10, 8))

plt.tight_layout()
plt.show()
```

```python
correlation = df[numerical_columns].corr()

correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Iris Features")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Species",
    y="PetalLengthCm"
)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")

plt.show()
```

### Species distribution

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Species",
    y="PetalWidthCm"
)

plt.title("Petal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Width (cm)")

plt.show()
```

### Feature distribution analysis using histogram

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Species",
    y="SepalLengthCm"
)

plt.title("Sepal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")

plt.show()
```

### Correlation analysis and heatmap

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Species",
    y="SepalWidthCm"
)

plt.title("Sepal Width Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

### Species-wise feature comparison using boxplot

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species"
)

plt.title("Relationship Between Petal Length and Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend(title="Species")

plt.show()
```
